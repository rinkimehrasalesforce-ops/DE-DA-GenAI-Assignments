```python
%sql
create catalog if not exists claims;

create schema if not exists claims.raw
comment 'schema for containing raw data';

create schema if not exists claims.bronze
comment 'schema for containing bronze data';

create schema if not exists claims.silver
comment 'schema for containing silver data';

create schema if not exists claims.gold
comment 'schema for containing gold data';
```

```python
%sql
create volume if not exists claims.raw.landing;
```

```python
dbutils.widgets.text("catalog", "claims")
catalog = dbutils.widgets.get("catalog")
```

#plans

```python
spark.sql(f"""
create or replace table {catalog}.raw.plans as
select * from read_files('/Volumes/claims/raw/landing/plans.csv')
""")
```

#hospitals

```python
spark.sql(f"""
create or replace table {catalog}.raw.hospitals as
select * from read_files('/Volumes/claims/raw/landing/hospitals.csv')
""")
```

#members

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {catalog}.raw.members AS
SELECT
    member_id,
    name,
    dob,
    gender,
    city,
    state,
    CAST(national_id AS STRING) AS national_id,
    phone,
    email,
    _rescued_data
FROM read_files('/Volumes/claims/raw/landing/members.csv')
""")
```

#hospital_network_history

```python
spark.sql(f"""
create or replace table {catalog}.raw.hospital_network_history as
select * from read_files('/Volumes/claims/raw/landing/hospital_network_history.csv')
""")
```
